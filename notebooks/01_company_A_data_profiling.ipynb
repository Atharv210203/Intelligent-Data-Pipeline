```python
import pandas as pd
import numpy as np
import os

BASE_PATH="../data/raw/Cantill_DataML_Task_Kit/company_A"

print(os.listdir(BASE_PATH))
```

```python
erp_df = pd.read_csv(os.path.join(BASE_PATH, "erp_goods_receipts.csv"))
tool_df = pd.read_csv(os.path.join(BASE_PATH, "invoice_tool_export.csv"))
labels_df = pd.read_csv(os.path.join(BASE_PATH, "labels_A.csv"))
taxonomy_df = pd.read_csv(os.path.join(BASE_PATH, "taxonomy_A.csv"))

print("ERP:", erp_df.shape)
print("Invoice Tool:", tool_df.shape)
print("Labels:", labels_df.shape)
print("Taxonomy:", taxonomy_df.shape)
```

```python
# Inspect ERP dataset
print("ERP DATASET")
erp_df.info()
```

```python
# Inspect Invoice Tool dataset
print("\nINVOICE TOOL DATASET")
tool_df.info()
```

```python
# Inspect Labels dataset
print("\nLABELS DATASET")
labels_df.info()
```

```python
# Inspect Taxonomy dataset
print("\nTAXONOMY DATASET")
taxonomy_df.info()
```

```python
display(erp_df.head())
display(tool_df.head())
display(labels_df.head())
display(taxonomy_df.head())
```

```python
# Missing values in each dataset
datasets={
    "ERP":erp_df,
    "Invoice Tool":tool_df,
    "Labels":labels_df,
    "Taxonomy":taxonomy_df
}

for name,df in datasets.items():
    print(f"\n{'='*40}")
    print(f"{name} - Missing Values")
    print(df.isnull().sum())


```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    missing_perc = (df.isnull().mean() * 100).round(2)
    print(missing_perc[missing_perc > 0].sort_values(ascending=False))
```

```python
# Check duplicate identifiers

print("ERP duplicate line IDs:", erp_df["line_id"].duplicated().sum())

print("Tool duplicate record IDs:",
      tool_df["tool_record_id"].duplicated().sum())

print("Labels duplicate record IDs:",
      labels_df["record_id"].duplicated().sum())

# Check complete duplicate rows
print("\nComplete duplicate rows:")
print("ERP:", erp_df.duplicated().sum())
print("Tool:", tool_df.duplicated().sum())
print("Labels:", labels_df.duplicated().sum())
```

```python
# ERP identifier statistics

print("ERP")
print("Total PO numbers:", len(erp_df["po_number"]))
print("Unique PO numbers:", erp_df["po_number"].nunique())
print("Unique vendors:", erp_df["vendor_id"].nunique())
print("Unique invoice references:", erp_df["invoice_reference"].nunique()
      if "invoice_reference" in erp_df.columns else "Column not available")

# Invoice Tool identifier statistics

print("\nINVOICE TOOL")
print("Unique invoice numbers:", tool_df["invoice_number"].nunique())
print("Total PO references:", len(tool_df["po_reference"]))
print("Unique PO references:", tool_df["po_reference"].nunique())
print("Unique supplier names:", tool_df["supplier_name_as_captured"].nunique())
```

```python
print("ERP - Most frequent PO numbers")
display(erp_df["po_number"].value_counts().head(15))

print("\nInvoice Tool - Most frequent PO references")
display(tool_df["po_reference"].value_counts().head(15))
```
